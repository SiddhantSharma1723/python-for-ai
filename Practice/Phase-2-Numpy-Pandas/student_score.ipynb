```python
import numpy as np
rng = np.random.default_rng(1)
a = rng.integers(40,101,20).reshape(5,4)
print(a)
print(a.mean(axis = 1))
print(a.mean(axis = 0))
row_average = a.mean(axis = 1)
print(np.argmax(row_average))
b = a < 50
print(a[b])
c = np.where(a < 50, 50,np.floor(a))
print(c)
d = np.array([5])
bonus = a + d
clipped = np.clip(0,100,bonus) 
print(clipped)
new_student = rng.integers(40,101,4)
v = np.vstack([a,new_student])
print(v)
v_1d = v.flatten()
print(v_1d)
v_2d = v_1d.reshape(6,4)
print(v_2d)
scores = v_2d.sum(1)
print(scores)
std = scores.std()
print(std)
```
