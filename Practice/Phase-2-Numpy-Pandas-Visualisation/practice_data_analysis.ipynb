```python
import pandas as pd
import matplotlib.pyplot as plt

pokemon = pd.read_csv("pokemon_stats_messy.csv")
# print(pokemon.info())
# print(pokemon.describe())
# print(pokemon.isnull().sum())
pokemon["HP"] = pd.to_numeric(pokemon["HP"],errors = "coerce")

pokemon["Type2"] = pokemon["Type2"].replace({
    "None": "Unknown"
})
pokemon["Type2"] = pokemon["Type2"].fillna("Unknown")
pokemon["Type1"] = pokemon["Type1"].astype(str).str.strip().str.title()

pokemon = pokemon.drop_duplicates()

q1 = pokemon["Attack"].quantile(0.25)
q3 = pokemon["Attack"].quantile(0.75)

IQR = q3 - q1

lower = q1 - 1.5*IQR
upper = q3 + 1.5*IQR
pokemon = pokemon[(pokemon["Attack"]>=lower) & (pokemon["Attack"]<=upper)]


'''Now Visualisation'''
attack_distribution = pokemon["Attack"].plot(kind = "hist",bins = 20)
pokemon.boxplot(column = "Defense",by = "Type1")
pokemon.boxplot(column = "Defense",by = "Legendary")

# selected1 = pokemon["Attack","Defense"]
# selected1graph = pd.plotting.scatter_matrix(pokemon[selected1])

# selected2 = pokemon["Speed","Sp_Atk"]
# selected2graph = pd.plotting.scatter_matrix(pokemon[selected2])
pd.plotting.scatter_matrix(pokemon[["Attack","Defense"]])
plt.savefig("Attack vs Defense")
pd.plotting.scatter_matrix(pokemon[["Speed","Sp_Atk"]])
plt.savefig("Speed vs Sp_Atk")


pokemon.info()

'''The Data shows how some values can be missing or mislabelled and gives the perfect example for how a messy data file looks like..
combining air spaces to different names for missing value.'''
```

```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots(2,1)
motivation = [45,35,25,48,15]
discipline = [10,20,30,40,50]
time = [100,200,300,400,500]
ax[0].plot(time,motivation,label = "Motivtion",color = "crimson",linestyle =(":"),marker = ("o"))
ax[1].plot(time,discipline,label = "Discipline",color = "blue",linestyle = ("--"),marker = ("v"))
ax[0].set_title("Is motivation better?")
ax[1].set_title("Or discipline.")
ax[0].set_xlabel("Time")
ax[1].set_xlabel("Time")
ax[0].legend()
ax[1].legend()

plt.tight_layout()
fig.savefig("Whats Better.png")
plt.show()
```
