```python
'''This is a Number Guessing Game'''

import random
guess = random.randint(1,100)

while True:
    your_guess = int(input("Enter your guess between 1 to 100: "))
    difference = abs(your_guess - guess)
    if your_guess < guess:
        if difference <= 10:
            print("You are close..but try a little higher.")
        else:
            print("Try higher.")
    elif your_guess > guess:
        if difference <= 10:
            print("You are close..but try a little lower.")
        else:
            print("Try lower.")
    else:
        print("Congratulations! You guessed the number correctly.")
        for i in range(10):
            print("You are a genius!")
        break
```

```python
import requests
import datetime as dt
import pandas as pd
today = dt.datetime.now()
week_ago = today - dt.timedelta(days = 7)

start_date = week_ago.strftime("%d-%m-%Y")
end_date = today.strftime("%d-%m-%Y")

url = f"https://api.open-meteo.com/v1/forecast?latitude=26.84&longitude=80.94&start={start_date}&end={end_date}&daily=temperature_2m_max,temperature_2m_min"
response = requests.get(url)
data = response.json()
print(data)

#-----------------------------------------

daily_data = data['daily']

df = pd.DataFrame({
    'date': daily_data['time'],
    'max_temp': daily_data['temperature_2m_max'],
    'min_temp': daily_data['temperature_2m_min']
})

df['date'] = pd.to_datetime(df['date'])

print(df)

#-----------------------------------------

import matplotlib.pyplot as plt

# Create the plot
plt.figure(figsize=(10, 6))
plt.plot(df['date'], df['max_temp'], marker='o', label='Max Temp')
plt.plot(df['date'], df['min_temp'], marker='o', label='Min Temp')

# Add labels and title
plt.xlabel('Date')
plt.ylabel('Temperature (°C)')
plt.title('Paris Weather - Past 7 Days')
plt.legend()

# Rotate x-axis labels for readability
plt.xticks(rotation=45)
plt.tight_layout()

# Save the plot
plt.savefig('weather_chart.png')
plt.show()
```

```python
class data_validator:
    def __init__(self):
        self.errors = []
    def email(self, email):
        if "@" not in email:
            self.errors.append("Invalid email address.")
            return False
        return True
    def phone_number(self, number):
        if len(number) != 10 or type(number) != int:
            self.errors.append("Invalid phone number.")
            return False
        return True
    def age(self, age):
        if age < 0 > 120:
            self.errors.append("Invalid age.")
            return False
        return True
validator = data_validator()
print(validator.email(email="siddhantsharma61594@gmail.com"))
print(validator.phone_number(number = "1234567895"))
print(validator.age(age=18))
print(validator.errors)
```

```python
class animal():
    def __init__(self, name):
        self.name = name
    def eat(self):
        return f"{self.name} is eating."
    def sleep(self):
        return f"{self.name} is sleeping."

class cat(animal):
    def meow(self):
        return f"{self.name} says Meow!"
    def bhau(self):
        return f"But {self.name} says Bhau!"
    
cat_1 = cat("Shishi")
cat_2 = cat("Mishi")
print(cat_1.eat())
print(cat_2.sleep())
print(cat_1.meow())
print(cat_2.bhau())
```

```python
a = int(input("Enter a number:"))
if a <= 1:
    print("The number is not a prime number")
else:
    prime = True
    for i in range(2, a):
        if a % i == 0:
            prime = False
            break
    if prime == True:
        print("the number is a prime number")
    else:
        print("The number is not a prime number")
```
